### Plotting KFC same-stores sales growth over time

This data on KFC's same-store sales growth was downloaded from [Source: Bloomberg Finance L.P.](https://docs.google.com/spreadsheets/d/16qerdsU3p2T1IpAABoSlj1EgwpQPjGlOzDxg-2RLNm0/edit?usp=sharing).

Here are the steps we need to take:
- Filter our data using a lambda function
- Format a data column into a datetime 
- Use the resample function
- plot the data with matplotlib

First import `pandas` and `matplotlib` (`import matplotlib.pyplot as plt`):

```python
import pandas as pd
import matplotlib.pyplot as plt
```

Read in the csv file with the sales growth data

```python
%%time
kfc_salesgrowth = pd.read_csv('../data/KFC_SameStoreSales_Growth.csv')
```

```python
kfc_salesgrowth.head()
```

```python
kfc_salesgrowth.dtypes
```

### Resampling our data over time
- format a column as a date time
- set the index of our dataframe to this date time column
- resample our dataframe 

```python
kfc_salesgrowth['Date'] = kfc_salesgrowth['Date'].astype('datetime64[ns]')
```

Check data type

```python
kfc_salesgrowth.dtypes
```

```python
kfc_salesgrowth = kfc_salesgrowth.set_index('Date')
kfc_salesgrowth.head()
```

### Plot the data 

See KFC quarterly same-store sales growth since 2021

```python
plt.plot(kfc_salesgrowth)
```

### Modifying the chart

Customizing the chart to better convey information to readers

```python
fig, ax = plt.subplots()  # Create a figure containing a single axes (using a different function)
ax.plot(kfc_salesgrowth)
```

### Label chart

```python
fig, ax = plt.subplots()
ax.plot(kfc_salesgrowth)
ax.set_xlabel('Quarter', fontsize=14)
ax.set_ylabel('Sales Growth (%)', fontsize=14)
ax.set_title('KFC Same Store Sales Growth', fontsize=14)
```

### Change size of chart

```python
fig, ax = plt.subplots()
fig.set_size_inches(9,6)
ax.plot(kfc_salesgrowth)

ax.set_xlabel('Reported Date', fontsize=14)
ax.set_ylabel('Year-over-year Sales Growth (%)', fontsize=14)
ax.set_title('KFC Same Store Sales Growth', fontsize=14)

plt.plot(kfc_salesgrowth)
plt.savefig('../outputs/chart.svg')
plt.savefig('../outputs/chart.png')
```
